```python
names = ['cora', 'netscience', 'power', 'euroroad', 'yeast', 'crime']
datasets = [
'FINAL_cora_res.csv',
'FINAL_netscience_res.csv',
'FINAL_power_res.csv',
'FINAL_euroroad_res.csv',
'FINAL_yeast_res.csv',
'FINAL_crime_res.csv']
data, res = ut.mix_dataset(datasets, 'Hits@50', filtre2=['evc', 'pr', 'csgcl'])

datasets_sbm = [
'FINAL_SBM_cora_res.csv',
'FINAL_SBM_netscience_res.csv',
'FINAL_SBM_power_res.csv',
'FINAL_SBM_euroroad_res.csv',
'FINAL_SBM_yeast_res.csv',
'FINAL_SBM_crime_res.csv']
data_sbm, res_sbm = ut.mix_dataset(datasets_sbm, 'Hits@50', filtre=['sbm\+'])

datasets_r = [
'R_FINAL_cora_res.csv',
'R_FINAL_netscience_res.csv',
'R_FINAL_power_res.csv',
'R_FINAL_euroroad_res.csv',
'R_FINAL_yeast_res.csv',
'R_FINAL_crime_res.csv']
data_r, res_r = ut.mix_dataset(datasets_r, 'Hits@50', filtre2=['rjc', 'rra'])

datasets_A = [
'A2_FINAL_cora_res.csv',
'A2_FINAL_netscience_res.csv',
'A2_FINAL_power_res.csv',
'A2_FINAL_euroroad_res.csv',
'A2_FINAL_yeast_res.csv',
'A2_FINAL_crime_res.csv']
data_a, res_a = ut.mix_dataset(datasets_A, 'Hits@50')

datasets_L = [
'L_FINAL_cora_res.csv',
'L_FINAL_netscience_res.csv',
'L_FINAL_power_res.csv',
'L_FINAL_euroroad_res.csv',
'L_FINAL_yeast_res.csv',
'L_FINAL_crime_res.csv'
]
data_l, res_l = ut.mix_dataset(datasets_L, 'Hits@50')

for df in [data, res, data_sbm, res_sbm, data_r, res_r, data_a, res_a, data_l, res_l]:
	df.index = names

for df in [data_sbm, data_r, data_a, data_l]:
	data = data.join(df)
	
for df in [res_sbm, res_r, res_a, res_l]:
	res = res.join(df)
big_tab = ut.tex_table(data, res).T
big_tab
```

```python
names = ['cora', 'citeseer', 'pubmed']
datasets = [
'FINAL_cora_res.csv',
'FINAL_citeseer_res.csv',
'FINAL_pubmed_res.csv',
]
data, res = ut.mix_dataset(datasets, 'Hits@100', filtre2=['csgcl', 'pr', 'evc'])
data.index = names
res.index = names

datasets_L = [
'L_FINAL_cora_res.csv',
'L_FINAL_citeseer_res.csv',
'L_FINAL_pubmed_res.csv',
]
data_l, res_l = ut.mix_dataset(datasets_L, 'Hits@100', filtre2=['csgcl', 'pr', 'evc'])
data_l.index = names
res_l.index = names

datasets_A = [
'A2_FINAL_cora_res.csv',
'A2_FINAL_citeseer_res.csv',
'A2_FINAL_pubmed_res.csv',
]
data_a, res_a = ut.mix_dataset(datasets_A, 'Hits@100', filtre2=['csgcl', 'pr', 'evc'])
data_a.index = names
res_a.index = names

for df in [data_a, data_l]:
	data = data.join(df)
	
for df in [res_a, res_l]:
	res = res.join(df)

big_tab = ut.tex_table(data, res).T
big_tab
```
