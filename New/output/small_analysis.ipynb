```python
SMALL_DATASETS = ["facebook_friends", "wiki_science", "crime", 
                  "power", "unicodelang", "euroroad", 
                  "pol_kato", "pol_robertson"]

for name in SMALL_DATASETS:
    result = dict()
    for ep in [0, 10, 30]:
        data = ut.res_load(f'ep{ep}_{name}_res.csv')
        _, data = ut.split_methods(data, '_mean')
        for method in data.index:
            if method not in result.keys():
                result[method] = []
                if ep == 10:
                    result[method] = [result['bgrl_random_mean'][0]]
            result[method].append(ut.extract_mean(data.loc[method, 'ROCAUC']))
    ut.plot_epochs(result, name)
ut.heatmap(data, 'ep0_crime_res.csv')
data = ut.res_load('ep10_crime_res.csv')
_, data = ut.split_methods(data, '_mean')
ut.heatmap(data, 'ep10_crime_res.csv')
data = ut.res_load('ep30_crime_res.csv')
_, data = ut.split_methods(data, '_mean')
ut.heatmap(data, 'ep30_crime_res.csv')
```

```python
# ROCAUC score
names = ['test_reconstruct_facebook_friends_res.csv',
'test_reconstruct_wiki_science_res.csv',
'test_reconstruct_crime_res.csv',
'test_reconstruct_unicodelang_res.csv',
'test_reconstruct_euroroad_res.csv',
'test_reconstruct_power_res.csv']
all_data, all_res = ut.mix_dataset(names, 'ROCAUC')
display(ut.tex_table(all_data).T)
all_data, all_res = ut.mix_dataset(names, 'AP')
display(ut.tex_table(all_data).T)
all_data, all_res = ut.mix_dataset(names, 'Hits@100')
display(ut.tex_table(all_data).T)
all_data, all_res = ut.mix_dataset(names, 'Hits@50')
display(ut.tex_table(all_data).T)
all_data, all_res = ut.mix_dataset(names, 'Hits@20')
display(ut.tex_table(all_data).T)
```

```python
all_data, all_res = ut.mix_dataset(names, 'ROCAUC')
all_res_t = all_res.stack().reset_index()
all_res_t.columns = ['data_name', 'methods', 'score']
all_res_t = all_res_t.explode('score')
all_res_t = all_res_t.reindex(columns=['methods', 'data_name', 'score'])
all_res_t['data_name'] = all_res_t['data_name'] + '_' + all_res_t.groupby(['data_name', 'methods']).cumcount().add(1).astype(str)
all_res_t.to_csv('example.csv', index=False)
```

```python
from scipy.stats import friedmanchisquare
from aeon.visualisation import plot_critical_difference
rankings_matrix = all_res.rank(axis=1, method='min', ascending=True)
friedman_stat, p_value = friedmanchisquare(*rankings_matrix.T.values)
print(f"Friedman test statistic: {friedman_stat}, p-value = {p_value}")
plot_critical_difference(all_res.T.values, all_res.columns)
```
